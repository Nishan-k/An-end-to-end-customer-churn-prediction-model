### 3.1 Check for duplicates:

```python
duplicates = df.duplicated()
print(duplicates.any(), "\n")
```

```python
print(f"There is a total of: {df[duplicates].shape[0]:,} duplicate rows. Original data has: {df.shape[0]:,} rows.")
```

Dropping the duplicates:

```python
df.drop_duplicates(inplace=True)
print(f"Number of rows after dropping the duplicates: {df.shape[0]:,}")
```

## 3.2 Handling Returned Quantity & Cancelled Orders:

```python
df.head(2)
```

```python
df['orders_returned'] = [1 if q < 0  else 0 for q in df['Quantity']]
print(f"A total of {df.value_counts('orders_returned').iloc[1]:,} orders were returned.")
```

```python
df['orders_cancelled'] = [1 if invoice.startswith('C') else 0 for invoice in df['Invoice']]
print(f"A total of {df.value_counts('orders_cancelled').iloc[1]:,} cancelled orders were found.")
```

#### Create a new column quantities to abs value:

```python
df['Quantity_abs'] = df['Quantity'].map(lambda q: np.abs(q))
```

## 3.3 Outlier Detection:

```python
df['Quantity_abs'].plot(kind='density', bins=100)
```

```python
numerical_cols = ['Quantity_abs', 'Price']
   
```

```python
Q1 = np.percentile(df['Quantity_abs'], 25)
Q3 = np.percentile(df['Quantity_abs'], 75)
IQR = Q3 - Q1
multiplier = 8.8
lower_bound = Q1 - multiplier * IQR
upper_bound = Q3 + multiplier * IQR
```
