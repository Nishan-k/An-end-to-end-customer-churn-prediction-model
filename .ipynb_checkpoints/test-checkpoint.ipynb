Here's your complete reference guide for the entire process.

---

## Step 1 — Data Scoping (What Data to Keep)

**Rule 1:** Drop everything after 2011-03-31 completely. It doesn't exist for you.

**Rule 2:** Your training data (for features) comes only from the observation window:
```
Observation Window: 2009-12-01 to 2010-12-31
```

**Rule 3:** Keep only customers with >= 2 purchases in the observation window. Customers with 1 or 0 purchases don't have enough behavioral history to be meaningful.

---

## Step 2 — Define the Churn Label

Use the churn window purely for labeling. No features come from here.
```
Churn Window: 2011-01-01 to 2011-03-31
```

Logic:
```
For each eligible customer (from Step 1):
  - Did they purchase in the churn window? → Churn = 0 (retained)
  - Did they NOT purchase in the churn window? → Churn = 1 (churned)
```

Important: The churn window rows never enter your feature matrix. They are used only to assign 0 or 1 to each customer.

---

## Step 3 — Aggregate to Customer Level

Your data is currently transaction-level (one row per invoice). You must collapse it to customer-level (one row per CustomerID). Everything from Step 3 onwards works on the observation window data only.

---

## Step 4 — Customer-Level Features

These are calculated per CustomerID from the observation window:

**RFM Features:**
```
Recency   — days between the customer's last purchase and 2010-12-31
Frequency — total number of orders the customer made
Monetary  — total revenue generated by the customer (Quantity x Price)
```

**Additional Behavioral Features:**
```
Tenure              — days between customer's first ever purchase and 2010-12-31
AOV                 — Average Order Value = Monetary / Frequency
customer_cancel_rate — customer's own cancelled orders / customer's total orders
```

---

## Step 5 — StockCode-Level Features (Product Risk Metrics)

This is a separate aggregation — calculated at the product level, not customer level.

**Step 5a — Build a StockCode-level table:**

For every StockCode in the observation window, calculate:
```
stockcode_cancel_rate = cancelled orders for that StockCode / total orders for that StockCode
stockcode_return_rate = returned orders for that StockCode / total orders for that StockCode
```

Result is a table like:
```
StockCode | stockcode_cancel_rate | stockcode_return_rate
----------|-----------------------|----------------------
SKU001    | 0.15                  | 0.10
SKU002    | 0.02                  | 0.01
SKU003    | 0.08                  | 0.05
```

**Step 5b — Summarize back to customer level:**

For each customer, look up the stockcode_cancel_rate and stockcode_return_rate of every product they ever bought, then summarize:
```
avg_stockcode_cancel_rate — mean of stockcode_cancel_rate across all products the customer bought
max_stockcode_cancel_rate — max of stockcode_cancel_rate across all products the customer bought
avg_stockcode_return_rate — mean of stockcode_return_rate across all products the customer bought
max_stockcode_return_rate — max of stockcode_return_rate across all products the customer bought
```

Quick example:
```
C001 bought SKU001 (0.15), SKU002 (0.02), SKU003 (0.08)

avg_stockcode_cancel_rate = (0.15 + 0.02 + 0.08) / 3 = 0.083
max_stockcode_cancel_rate = 0.15
```

---

## Step 6 — Join Everything Together

You now have three separate tables. Join them all on CustomerID:

```
Customer-level features (Step 4)
        +
StockCode summaries per customer (Step 5b)
        +
Churn label (Step 2)
        ↓
Final dataframe: one row per customer, all features + churn label
```

---

## Final Feature Set Summary

```
CustomerID                   — identifier (drop before modeling)
Recency                      — days since last purchase
Frequency                    — total orders
Monetary                     — total revenue
Tenure                       — days since first purchase
AOV                          — average order value
customer_cancel_rate         — customer's own cancellation behavior
avg_stockcode_cancel_rate    — avg product-level cancel risk for products they bought
max_stockcode_cancel_rate    — max product-level cancel risk for products they bought
avg_stockcode_return_rate    — avg product-level return risk for products they bought
max_stockcode_return_rate    — max product-level return risk for products they bought
Churn                        — 0 (retained) or 1 (churned) ← target variable
```

---

## Key Principles to Remember

- **No data leakage** — features only from observation window, label only from churn window. These two windows never mix.
- **Unit of analysis is the customer** — every feature must reduce to a single scalar value per CustomerID.
- **StockCode metrics need two steps** — product-level first, then summarize to customer-level. Never try to attach product-level metrics directly to a customer without summarizing.
- **Churn window rows are invisible to the model** — they exist only to generate the label column.

This is your complete reference. Everything else from here is standard ML pipeline work.

````python
import os
from datetime import datetime
from openai import OpenAI
import streamlit as st

# Initialize OpenAI client (use environment variable or Streamlit secrets)
api_key = os.getenv("OPENAI_API_KEY") or st.secrets.get("OPENAI_API_KEY")
if not api_key:
    raise ValueError("OPENAI_API_KEY not found")
client = OpenAI(api_key=api_key)
MODEL = 'gpt-4o-mini'

def system_prompt(report_type=None, audience=None, include_recommendations=True):
    """
    System prompt for the LLM.
    """
    prompt = f"""
You are a **Customer Retention Analyst** AI. Your task is to generate a concise, actionable report explaining customer churn risk based on SHAP values and model predictions.

**Rules**:
- Positive SHAP values **increase** churn risk; negative SHAP values **decrease** churn risk.
- Focus on the top 3-5 features with the highest absolute SHAP values.
- Express SHAP values as percentages (e.g., 0.05 = 5% impact).
- Base all insights strictly on the provided data.

**Output Structure**:
- **Title**: "Customer Churn Risk Report"
- **Prediction**: State the prediction and probability (e.g., "High Risk – 62% chance of churn").
- **Top Drivers**: List the most impactful features, explaining whether they increase or decrease risk, and by how much.
- **Business Interpretation**: Explain what each driver means in practical business terms.
"""
    if include_recommendations:
        prompt += """
- **Recommendations**: Suggest 2-3 specific, actionable interventions based on the top drivers. Each recommendation should include estimated impact (high/medium/low) and implementation difficulty (easy/moderate/difficult).
"""
    if report_type == "Executive Summary":
        prompt += "\nKeep the report very brief, focusing on business impact and ROI. Avoid technical details."
    elif report_type == "Detailed Analysis":
        prompt += "\nProvide deeper explanations, more context, and supporting details. Include possible underlying causes."
    elif report_type == "Technical Deep Dive":
        prompt += "\nInclude technical details about SHAP values, feature engineering, and model behavior. Use precise data science terminology."
    elif report_type == "Action Plan":
        prompt += "\nFormat the report as a step-by-step action plan with specific actions, responsible roles, and success metrics."

    if audience == "Management":
        prompt += "\nUse business language, highlight financial implications and strategic recommendations."
    elif audience == "Customer Service Team":
        prompt += "\nFocus on customer experience, practical guidance for interactions, and service improvements."
    elif audience == "Technical Team":
        prompt += "\nUse technical language, discuss implementation details and system improvements."
    elif audience == "Marketing Team":
        prompt += "\nFocus on customer segmentation, messaging, and campaign ideas to reduce churn."

    return prompt

def user_prompt(results, report_type=None, audience=None, include_recommendations=True):
    """
    User prompt containing the actual data.
    results: dict from API response containing 'prediction_results' and 'shap_values'
    """
    pred = results['prediction_results'][0]
    prediction = pred['prediction']
    prob_churn = pred['prediction_probability']
    risk_label = pred['risk_label']
    
    shap_list = results['shap_values']  # list of dicts: feature, value, shap_value
    
    # Build dictionaries for SHAP values and customer feature values
    shap_dict = {item['feature']: item['shap_value'] for item in shap_list}
    feature_values = {item['feature']: item['value'] for item in shap_list}
    
    # Sort by absolute SHAP value
    sorted_shap = dict(sorted(shap_dict.items(), key=lambda x: abs(x[1]), reverse=True))
    
    prompt = f"""
Generate a customer churn report based on the following data:

**Prediction**: {prediction}
**Churn Probability**: {prob_churn:.2%}
**Risk Label**: {risk_label}

**Feature values (customer data)**: {feature_values}

**SHAP values** (positive = increases churn risk, negative = decreases risk):
{sorted_shap}

**Current date**: {datetime.now().strftime('%B %d, %Y')}
"""
    if report_type:
        prompt += f"\n**Report Type**: {report_type}"
    if audience:
        prompt += f"\n**Target Audience**: {audience}"
    if not include_recommendations:
        prompt += "\n**Do NOT include recommendations** in this report."

    prompt += """
\nInterpretation guidelines:
- Positive SHAP → increases churn risk; negative → decreases.
- Use absolute SHAP values to rank importance.
- Convert SHAP values to percentages (e.g., 0.05 = 5% impact).
- Keep the report clear and actionable based on the data provided.
"""
    return prompt

def get_report(results, report_type=None, audience=None, include_recommendations=True):
    """
    Stream the report from OpenAI.
    """
    system = system_prompt(report_type, audience, include_recommendations)
    user = user_prompt(results, report_type, audience, include_recommendations)
    
    messages = [
        {"role": "system", "content": system},
        {"role": "user", "content": user}
    ]
    
    report_placeholder = st.empty()
    full_response = ""
    
    stream = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        stream=True
    )
    
    for chunk in stream:
        if chunk.choices[0].delta.content:
            full_response += chunk.choices[0].delta.content
            report_placeholder.markdown(full_response)
    
    return full_response
````
